```python
tradeiqx_instructions = """
You are TradeIQX, an AI trading mentor.

Your primary purpose is to educate users about:
- Stock markets
- Investing
- Technical analysis
- Fundamental analysis
- Risk management
- Paper trading
- Trading psychology

PERSONALITY:
- Friendly
- Clear
- Patient
- Encouraging
- Professional but easy to understand

TEACHING STYLE:
- Explain difficult concepts in simple language.
- Assume the user is a beginner unless their questions
  indicate otherwise.
- Give simple examples when useful.
- Define technical terms before using them.
- For complex questions, explain the answer step-by-step.

SAFETY:
- TradeIQX is an educational tool.
- Never guarantee profits or claim that a stock will
  definitely rise or fall.
- Do not present predictions as certain facts.
- Explain risks when discussing trading strategies.
- Encourage users to do their own research.

When the user asks a question, answer it directly first,
then provide additional explanation if useful.
"""
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="What is RSI?",
    config={
        "system_instruction": tradeiqx_instructions
    }
)

print(response.text)
```

```python
chat = client.chats.create(
    model="gemini-3.5-flash-lite",
    config={
        "system_instruction": tradeiqx_instructions
    }
)

print("TradeIQX AI Mentor is ready!")
print("Type 'bye' to exit.")
```

```python
while True:
    user_message = input("You: ")

    if user_message.lower() == "bye":
        print("TradeIQX: Goodbye! Keep learning. 📈")
        break

    response = chat.send_message(user_message)

    print("TradeIQX:", response.text)
```

```python
def ask_tradeiqx(question):

    # Step 1: Retrieve relevant TradeIQX knowledge
    context = tradeiqx_retriever.build_rag_context(question)

    # Step 2: Combine retrieved knowledge with the question
    prompt = f"""
Use the following TradeIQX knowledge to answer the user's question.

TRADEIQX KNOWLEDGE:
{context}

USER QUESTION:
{question}

Instructions:
- Use the provided TradeIQX knowledge as your primary source.
- Explain the answer clearly and simply.
- Do not invent facts that contradict the provided knowledge.
- If the knowledge does not contain enough information,
  say so.
- Give educational information, not financial advice.
"""

    # Step 3: Ask Gemini
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt,
        config={
            "system_instruction": tradeiqx_instructions
        }
    )

    return response.text
```

```python
print(ask_tradeiqx("What is RSI?"))
```

```python
print(ask_tradeiqx("What are option stocks"))
```

```python
print(ask_tradeiqx("how to make a cake?"))
```
